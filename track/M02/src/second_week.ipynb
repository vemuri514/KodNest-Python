```python
#Check whether number is positive ,negative or zero
# Read the number
number = int(input())

# Check whether the number is positive, negative or zero
if number > 0:
    print("Number is Positive")
elif number < 0:
    print("Number is Negative")
else:
    print("Number is Zero")
```

```python
#student eligibility checker
# Read marks, attendance, and project completion status
marks = int(input())
attendance = int(input())
project_completion = input()

# Check academic requirements using a compound condition
if marks >= 60 and attendance >= 75:
    # Check project completion using a nested condition
    if project_completion == "yes":
        print("Eligible")
    else:
        print("Not Eligible")
else:
    print("Not Eligible")
```

```python
# Clculate total using while loop
# Read the value of n
n = int(input())
# Initialize the counter and total
counter = 1
total = 0

# Calculate the total using a while loop
while counter <= n:
    total += counter
    counter += 1

# Display the total
print(f"Total: {total}")
```

```python
#Display a number sequence and characters
# Read the number and word
N = int(input())
word = input()

# Print the number sequence
print("Numbers:")
for i in range(1, N + 1):
    print(i)

# Print the characters
print("Characters:")
for char in word:
    print(char)
```

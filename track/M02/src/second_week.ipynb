```python
# MODULE 2
#-----------
# Create the required variables
#-----------------------------
student_name = "Aarav"
course_duration = 30
course_rating = 4.8
is_enrolled = True

# Display the data type of each variable
print(type(student_name))
print(type(course_duration))
print(type(course_rating))
print(type(is_enrolled))
```

```python
# Initial variables
#-------------------
age_text = "21"
course_fee_text = "499.50"
attempts = 3

# Convert the values
age = int(age_text)
course_fee = float(course_fee_text)
attempts_text = str(attempts)

# Display the converted values and their data types
print(age)
print(type(age))
print(course_fee)
print(type(course_fee))
print(attempts_text)
print(type(attempts_text))
```

```python
#Student age calculation
#--------------------------
# Original string variable
current_age = "21"
# Convert to integer and add 1
next_age = int(current_age) + 1
# Print the result using an f-string
print(f"Age next year: {next_age}")
```

```python
# Read and convert the student details
#------------------------------------
student_name = input()
student_age = int(input())
course_rating = float(input())

# Display the values using f-strings
print(f"Student: {student_name}")
print(f"Age: {student_age}")
print(f"Rating: {course_rating}")

# Display the data types
print(type(student_name))
print(type(student_age))
print(type(course_rating))
```

```python
#Division and Floor division
# Read input values
first_number = int(input())
second_number = int(input())

# Calculate and print regular division and floor division
print(first_number / second_number)
print(first_number // second_number)
```

```python
# Read the numbers
first_number = int(input())
second_number = int(input())

# Calculate the sum
sum_result = first_number + second_number

# Display the result
print(f"Sum: {sum_result}")
```

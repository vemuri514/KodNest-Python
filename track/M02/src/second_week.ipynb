```python
#student eligibility checker
# Read marks, attendance, and project completion status
marks = int(input())
attendance = int(input())
project_completion = input()

# Check academic requirements using a compound condition
if marks >= 60 and attendance >= 75:
    # Check project completion using a nested condition
    if project_completion == "yes":
        print("Eligible")
    else:
        print("Not Eligible")
else:
    print("Not Eligible")
```

```python
# Clculate total using while loop
# Read the value of n
n = int(input())
# Initialize the counter and total
counter = 1
total = 0

# Calculate the total using a while loop
while counter <= n:
    total += counter
    counter += 1

# Display the total
print(f"Total: {total}")
```

```python
# Analyze multiples of three
limit = int(input())
target = int(input())

count = 0
total = 0
found = False

# Examine every number from 1 to the limit
for i in range(1, limit + 1):
    if i % 3 == 0:
        count += 1
        total += i
        if target == i:
            found = True

target_found = 'Yes' if found else 'No'

# Display the count, total and search result
print("Count:", count)
print("Sum:", total)
print(f"Target Found: {target_found}")
```

```python
# Even number
# Read the limit
limit = int(input())

# Initialize the loop variable and total
number = 1
total = 0

# Examine every number from 1 to limit
while number <= limit:
    if number % 2 == 0:
        total += number
    number += 1

# Display the result
print(f"Even Sum: {total}")
```

```python
# Analyze entered numbers
# Read total number of inputs
number_count = int(input())

# Initialize counters and total accumulator
positive_count = 0
negative_count = 0
zero_count = 0
total = 0

# Process each number one by one
for _ in range(number_count):
    number = int(input())
    total += number

    if number > 0:
        positive_count += 1
    elif number < 0:
        negative_count += 1
    else:
        zero_count += 1

# Display final analysis
print(f"Positive Count: {positive_count}")
print(f"Negative Count: {negative_count}")
print(f"Zero Count: {zero_count}")
print(f"Total: {total}")
```

```python
# Student marks summary
# Read the total number of students
student_count = int(input())

# Initialize accumulators and counters
total_marks = 0
passed_count = 0
failed_count = 0

# Read and process each student's mark
for _ in range(student_count):
    mark = int(input())
    total_marks += mark
    
    if mark >= 40:
        passed_count += 1
    else:
        failed_count += 1

# Display summary results
print(f"Total Marks: {total_marks}")
print(f"Passed Students: {passed_count}")
print(f"Failed Students: {failed_count}")

# Display batch evaluation
if failed_count == 0:
    print("Batch Result: All Passed")
else:
    print("Batch Result: Needs Improvement")
```
